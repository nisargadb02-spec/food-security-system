```python
import pandas as pd

df = pd.read_csv("../data/hunger_india.csv")
df
```

```python
import matplotlib.pyplot as plt

plt.plot(df["Year"], df["Undernourished_Million"])
plt.title("Undernourished Population in India")
plt.xlabel("Year")
plt.ylabel("Millions")
plt.show()
```

```python
plt.plot(df["Year"], df["Poverty_Rate"])
plt.title("Poverty Rate in India")
plt.xlabel("Year")
plt.ylabel("Percentage")
plt.show()
```

```python
print("Key Insights:")
print("- Undernourished population is decreasing over time")
print("- Poverty rate shows gradual improvement")
print("- Both indicators are correlated and improving slowly")
```
